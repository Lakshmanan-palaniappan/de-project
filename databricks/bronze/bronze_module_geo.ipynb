```python
from pyspark.sql.functions import *

bronze_df = spark.read.format("parquet").load(
    "abfss://bronze@travelappprojectstorage.dfs.core.windows.net/geofence_alert"
)
display(bronze_df)
bronze_df.printSchema()
print(f"Bronze Record Count: {bronze_df.count()}")
print("Null AlertId:",
      bronze_df.filter(col("AlertId").isNull()).count())
print("Null TouristId:",
      bronze_df.filter(col("TouristId").isNull()).count())
print("Null PlaceId:",
      bronze_df.filter(col("PlaceId").isNull()).count())
print("Null CreatedAt:",
      bronze_df.filter(col("CreatedAt").isNull()).count())
print("Negative Distance:",
      bronze_df.filter(col("DistanceMeters") < 0).count())
```
