set the path for module 2 working

```python
base_path = "abfss://bronze@travelappprojectstorage.dfs.core.windows.net/"
```

reading the tables inside the bronze layer

```python
df_alert = spark.read.parquet(base_path + "geofence_alert/")
df_place = spark.read.parquet(base_path + "danger_place/")
df_tourist = spark.read.parquet(base_path + "tourist/")
df_agency = spark.read.parquet(base_path + "agency/")
df_trip = spark.read.parquet(base_path + "trip/")
```

checking the data inside the tables in bronze layer


```python
display(df_alert)
display(df_place)
display(df_tourist)
display(df_agency)
display(df_trip)
```

checking the schema of the tables

```python
df_alert.printSchema()
df_place.printSchema()
df_tourist.printSchema()
df_agency.printSchema()
df_trip.printSchema()
```

count of records inside each file

```python
print("geofence_alert:", df_alert.count())
print("danger_place:", df_place.count())
print("tourist:", df_tourist.count())
print("agency:", df_agency.count())
print("trip:", df_trip.count())
```
