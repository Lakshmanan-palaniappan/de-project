```python
from pyspark.sql.functions import *

bronze_df = spark.read.format("parquet").load(
    "abfss://bronze@travelappprojectstorage.dfs.core.windows.net/latest_trilateration"
)
display(bronze_df)
bronze_df.printSchema()
print(
    f"Bronze Record Count : {bronze_df.count()}"
)
null_count = bronze_df.filter(
    col("TouristId").isNull()
).count()
print(
    f"Null TouristId Count : {null_count}"
)

null_updatedat_count = bronze_df.filter(
    col("UpdatedAt").isNull()
).count()

print(
    f"Null UpdatedAt Count : {null_updatedat_count}"
)

invalid_accuracy_count = bronze_df.filter(
    col("Accuracy") <= 0
).count()

print(
    f"Invalid Accuracy Count : {invalid_accuracy_count}"
)
```
